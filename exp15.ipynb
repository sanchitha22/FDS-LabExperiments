```python
import pandas as pd
import matplotlib.pyplot as plt

# Data collection
data = {
    'Name': ['Arun', 'Priya', 'Kavin', 'Divya', 'Ravi'],
    'Age': [18, 19, 18, 20, 19],
    'Marks': [75, 85, 60, 90, 70],
    'Attendance': [80, 95, 70, 98, 75]
}

# Create DataFrame
df = pd.DataFrame(data)

# Initial exploration
print("Student Dataset:")
print(df)

print("\nFirst 5 Records:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
print(df.describe())

# Visualization
plt.bar(df['Name'], df['Marks'])
plt.title('Student Marks Analysis')
plt.xlabel('Student Name')
plt.ylabel('Marks')
plt.show()
```
